my_dict={'name':['Yash','Neel','Dev'],'rollno':[11,12,13],'marks':[78,56,98]}
Perform following operations on above dictionary
a) Display name as Dev
c) Display greatest marks
b) Display 12 roll no

```python
# Create a dcitionary 
my_dict = {
    'name': ['Yash', 'Neel', 'Dev'],
    'rollno': [11, 12, 13],
    'marks': [78, 56, 98]
}
```

```python
# Subpart a. Displaying name as Dev

a = my_dict['name'][2]
print(a)
```

```python
# Subpart b. Diplaying greatest marks

b = max(my_dict['marks'])
print(b)
```

```python
# Subpart c. Displaying 12 rollno

c = my_dict['rollno'][1]
print(c)
```

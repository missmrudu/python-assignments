## Question 7
Calculate the frequency of letters occuring in a given string 


```python
# Calculating the frquency of the letter 'a' in the given string

count = 0
str1 = "Hi, I am John. I am learning Python programming language. Python is a great language for data science and machine learning."
for ch in str1:
    if ch == 'a':
        count += 1
print(count)
```

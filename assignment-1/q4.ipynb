## Question 4 
Convert a dictionary to lists. 
Make one list of keys and the other of values

```python
# Create a dictionary to make lists of keys and values

my_dict = {
    'Name': 'James Bond',
    'Age': 23,  
    'Branch': 'CSE',
    'Specialization': 'Core',
    'Division': 'A'
}
```

```python
# Make list of keys and values from the dictionary

list_keys = list(my_dict.keys())
list_values = list(my_dict.values())

```

```python
# Printing the lists of keys and values

print("list of keys:", list_keys)
print("list of values:", list_values)
```

Sort the elements of the dictionary using sorted() function
my_dict={'name':'ABC', 'panel'='B','rollno'=34, 'marks':[65,87,67,94]}
sorted(my dict)

```python
# Defining the dictionary 

my_dict = {
    'name': 'James Bond',
    'panel': 'B',
    'rollno': 34,
    'marks': [65, 87, 67, 94]
}
```

```python
# Using sorted function on the dictionary 

sorted_dict = sorted(my_dict)
```

```python
# printing the output

print(sorted_dict)
```

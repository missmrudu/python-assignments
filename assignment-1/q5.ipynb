## Question 5 
Find the mean of values of dictionary
    e.g mydict={'marks1':23,'marks2': 123, 'marks3': 43, 'marks4': 13, 'marks5': 39}
    output: mean of all values.

```python
import statistics

# Creating a dictionary 

my_dict = {
    'marks1':23,
    'marks2': 123,
    'marks3': 43,
    'marks4': 13,
    'marks5': 39
}
```

```python
# Using mean function to calculate the mean of the values in the dictionary

mean_values = mean(my_dict.values())
print(mean_values)
```

```python
# Using the sum and len functions to calculate the mean of the values in the dictionary

mean_values = sum(my_dict.values()) / len(my_dict)
print(mean_values)
```

## Question 9
Suppose a date is represented as a tuple (d,m.y). Write a program to create two date tuples
with values taken from the user and find the number of days between the two dates as an output.

```python
# importing  datetime 

import datetime
 
```

```python
# Take d1 and d2 from the user in the format (d,m,y)

d1 = list(map(int, input("Enter the first date (dd/mm/yyyy): ").split("/")))
d2 = list(map(int, input("Enter the second date (dd/mm/yyyy): ").split("/")))
```

```python
# convert d1 and d2 to datetime objects

d1_date = datetime.date(int(d1[2]), int(d1[1]), int(d1[0]))
d2_date = datetime.date(int(d2[2]), int(d2[1]), int(d2[0]))
```

```python
#print the difference between d1 and d2 in days

print(d2_date - d1_date)
```

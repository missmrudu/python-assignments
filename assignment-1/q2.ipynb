## Question 2
Convert list to dictionary
    e.g. list1=['name', 'panel','rollno']
    list2=[‘ABC','B',34]
    my_dict= {'name':'ABC','panel'='B','rollno'=34}

```python
# Defining the lists

list1 = [
    'name',
    'panel',
    'rollno'
]

list2 = [
    'ABC',
    'B',
    34
]
```

```python
# Using zip function to make a dictionary using the defined lists

my_dict = dict(zip(list1, list2))
```

```python
# Printing the output dictionary
print(my_dict)
```

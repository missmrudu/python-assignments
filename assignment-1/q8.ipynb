## Question 8
Write a program which takes numerator and denominator values from the user and gives
output as a tuple of Quotient and Remainder.

```python
# Take numerator and denominator as input from the user 
 
num = int(input("Enter the numerator: "))
deno = int(input("Enter the denominator: "))
```

```python
# Using divmod function to calculate the quotient and remainder

divmod_result = divmod(num, deno)
```

```python
# Printing the result in a tuple

print(divmod_result)
```

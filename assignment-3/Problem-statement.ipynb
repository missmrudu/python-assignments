## To check whether or not the triangle is right-angled triangle using function


```python
#define a function that takes three sides of a triangle as parameters and checks if the triangle can be a right-angled triangle 

def right_triangle(a, b, c):
    # Sort the sides to identify the hypotenuse
    sides = sorted([a, b, c])
    hypotenuse = sides[2]
    side1 = sides[0]
    side2 = sides[1]

    # Check the Pythagorean theorem
    if hypotenuse**2 == side1**2 + side2**2:
        return True
    else:
        return False    
```

```python
#Call the function with example values which gives true 

right_triangle(3, 4, 5)
```

```python
#Call the function with example values which gives false

right_triangle(6,5,8)
```

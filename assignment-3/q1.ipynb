## Accept the number from the user and display prime factors of that no.  
### e.g. Input number=100 
### prime factors=2,5

```python
#Accept the number from the user
num = int(input("Input number = "))
```

```python
# Store unique prime factors
prime_factors = []
original_num = num
```

```python
# Start with the smallest prime number
divisor = 2
```

```python
while num > 1:
    if num % divisor == 0:
        # If it's a factor, add it to our list if it's not already there
        if divisor not in prime_factors:
            prime_factors.append(divisor)
        # Keep dividing the number to strip out this factor
        num //= divisor
    else:
        # Move to the next potential divisor
        divisor += 1
```

```python
# Display the output format
factors_str = ",".join(map(str, prime_factors))
print(f"prime factors = {factors_str}")
```

## Write a python program which accepts three coefficients of the quadratic equation and generates real roots (with same or different values). If (b 2-4ac) <0, use of math.sqrt function generates an exception as Value error"math domain error". Show exception handling for ValueFrror. NameError and TypeError.

```python
import math

def solve_quadratic():
    try:
        # Prompting user for coefficients
        # A NameError will trigger if a variable name inside the formula is mistyped.
        # A TypeError will trigger if we mix incompatible data types.
        a = float(input("Enter coefficient a: "))
        b = float(input("Enter coefficient b: "))
        c = float(input("Enter coefficient c: "))
        
        # Check if the equation is validly quadratic
        if a == 0:
            raise ValueError("Coefficient 'a' cannot be zero in a quadratic equation.")
            
        # Calculate the discriminant
        discriminant = b**2 - 4*a*c
        
        # This math.sqrt calculation will throw a ValueError: math domain error if discriminant < 0
        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)
        
        # Displaying the results
        if discriminant == 0:
            print(f"\nThe equation has one real repeated root:\nRoot = {root1}")
        else:
            print(f"\nThe equation has two distinct real roots:\nRoot 1 = {root1}\nRoot 2 = {root2}")
            
    except ValueError as e:
        # Catches negative square roots (math domain error) or custom bad values (a=0)
        print(f"\n[ValueError Caught]: {e}")
        print("Reason: You cannot take the square root of a negative number in real number systems.")
        
    except TypeError as e:
        # Catches internal datatype mismatches or operation mismatches
        print(f"\n[TypeError Caught]: {e}")
        print("Reason: An operation was performed on an incompatible data type.")
        
    except NameError as e:
        # Catches calls to variables or functions that do not exist or are misspelled
        print(f"\n[NameError Caught]: {e}")
        print("Reason: A variable or function name used in the script is not defined.")
```

```python
# Run the program
if __name__ == "__main__":
    solve_quadratic()
```
